```python
import folium
```

```python
# CONFIGURACIÓN DEL MAPA
# Coordenadas centrales (Aproximadamente frente a Fisterra/Coruña)
mapa_centro = [43.0, -9.5]
mapa = folium.Map(location=mapa_centro, zoom_start=8, tiles='OpenStreetMap')

# DIBUJAMOS LA ZONA DE VIGILANCIA (ALCANCE DEL RADAR)
# Representamos visualmente el área que monitoriza tu IA
folium.Circle(
    location=mapa_centro,
    radius=150000, # En metros (ejemplo visual de alcance)
    color='blue',
    fill=True,
    fill_color='blue',
    fill_opacity=0.05,
    popup='Zona de Cobertura VIGÍA-IA (ZEE)'
).add_to(mapa)
```

```python
# DATOS DE LOS BARCOS (Simulando lo que sale del modelo de IA)
barcos = [
    {
        "nombre": "Mercante 'Galicia Spirit'",
        "lat": 43.5,
        "lon": -9.8,
        "tipo": "green", # Verde = Normal
        "info": "Estado: NORMAL<br>Velocidad: 18 nudos<br>Rumbo: 180°"
    },
    {
        "nombre": "ALERTA: Embarcación Desconocida",
        "lat": 42.9,
        "lon": -9.3, # Cerca de Costa da Morte
        "tipo": "red", # Rojo = Alerta
        "info": "<b>!!! ALERTA DETECTADA !!!</b><br>"
                "Motivo: Parada injustificada en alta mar<br>"
                "Probabilidad Anomalía: 98%<br>"
                "Acción: Enviar Drone / Aviso S. Marítimo"
    }
]
```

```python
# AÑADIMOS LOS MARCADORES AL MAPA
for barco in barcos:
    # Elegimos el icono según el estado
    icono = 'info-sign' if barco['tipo'] == 'green' else 'exclamation-sign'

    folium.Marker(
        location=[barco['lat'], barco['lon']],
        popup=folium.Popup(barco['info'], max_width=300),
        tooltip=barco['nombre'], # Texto al pasar el ratón por encima
        icon=folium.Icon(color=barco['tipo'], icon=icono)
    ).add_to(mapa)

# GUARDAMOS EL RESULTADO
nombre_archivo = "tablero_vigia_galicia.html"
mapa.save(nombre_archivo)

print(f"Mapa generado con éxito: {nombre_archivo}")
print("Abre este archivo en tu navegador para ver la simulación.")
```
