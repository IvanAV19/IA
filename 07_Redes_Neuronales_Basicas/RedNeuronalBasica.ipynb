##**Función de activación sigmoide**

```python
def sigmoide(x):
    return 1/(1+np.exp(-x))
```

```python
xs=np.linspace(-10,10,100)
ys=sigmoide(xs)
plt.plot(xs,ys)
plt.plot([0,0],[0,1])
```

##**Funcion que ejecuta el procesamiento de una capa**

```python
def ejecutar_capa(pesos,sesgos,entradas):
    return sigmoide(np.dot(pesos,entradas)+sesgos)
```

```python
#probamos a ejecutar la primera capa con el primer dato de entrada
salidasCapa1=ejecutar_capa(pesosCapa1,sesgosCapa1,x[0])
print(salidasCapa1)
```

## **Al final nos quedamos con las siguientes formulas para calcular la derivada del error con respecto a cada parametro**
Captura desde 2025-11-07 12-27-35.png

```python
#programamos la función deltaS
def deltaS(y,prediccion):
    return (prediccion-y) * prediccion*(1-prediccion)

#Estas son las formulas de las derivadas con respecto a cada parametro de la red

#LOS NOMBRES DE LAS FUNCIONES SE CORRESPONDEN CON LAS POSICIONES DE LOS PARAMETROS EN LAS MATRICES
#NO CON LOS NOMBRES QUE APARECEN EN EL GRAFICO DE LA RED.
#EN EL COMENTARIO PREVIO A CADA FUNCIÓN SE EXPLICA A QUE PARAMETRO DEL GRAFICO SE CORRESPONDE

#Peso p11 en el grafico de la red
def der_p00(y,prediccion,pesos,salidas,entradas):
    return deltaS(y,prediccion)* pesos[0] * salidas[0]*(1-salidas[0])*entradas[0]

#Peso p13 en el grafico de la red
def der_p01(y,prediccion,pesos,salidas,entradas):
    return deltaS(y,prediccion)* pesos[0] * salidas[0]*(1-salidas[0])*entradas[1]

#Sesgo11 en el grafico de la red
def der_sesgo0(y,prediccion,pesos,salidas,entradas):
    return deltaS(y,prediccion)* pesos[0] * salidas[0]*(1-salidas[0])

#Peso p12 en el grafico de la red
def der_p10(y,prediccion,pesos,salidas,entradas):
    return deltaS(y,prediccion)* pesos[1] * salidas[1]*(1-salidas[1])*entradas[0]

#Peso p14 en el grafico de la red
def der_p11(y,prediccion,pesos,salidas,entradas):
    return deltaS(y,prediccion)* pesos[1] * salidas[1]*(1-salidas[1])*entradas[1]

#Sesgo12 en el grafico de la red
def der_sesgo1(y,prediccion,pesos,salidas,entradas):
    return deltaS(y,prediccion)* pesos[1] * salidas[1]*(1-salidas[1])

#Peso p21 en el grafico de la red
def der_p0(y,prediccion,salidas):
    return deltaS(y,prediccion)* salidas[0]

#Peso p22 en el grafico de la red
def der_p1(y,prediccion,salidas):
    return deltaS(y,prediccion)* salidas[1]

#Sesgo21 en el grafico de la red
def der_sesgoF(y,prediccion):
    return deltaS(y,prediccion)
```

```python
#Funcion de inferencia
def predecir(X,sesgoscapa1,sesgocapa2,pesoscapa1,pesoscapa2):
    salidascapa1=ejecutar_capa(pesoscapa1,sesgoscapa1,X)
    salida=ejecutar_capa(pesoscapa2,sesgocapa2,salidascapa1)
    return salidascapa1,salida
```

```python
#funcion del erro cuadratico medio
def EC(y,prediccion):
    return (y-prediccion)**2
```

```python
#Proceso de aprendizaje por back propagation
alfa=0.1
epocas=5000

for ep in range(epocas):
    error=0.0
    for i in range(len(x)):

        salidascapa1,prediccion=predecir(x[i],sesgosCapa1,sesgosCapa2,pesosCapa1,pesosCapa2)

        #aplicamos el ajuste con el alfa y las derivadas de cada parametro
        pesosCapa1[0,0]=pesosCapa1[0,0]-alfa*der_p00(y[i],prediccion,pesosCapa2,salidascapa1,x[i]).item()
        pesosCapa1[0,1]=pesosCapa1[0,1]-alfa*der_p01(y[i],prediccion,pesosCapa2,salidascapa1,x[i]).item()
        sesgosCapa1[0]=sesgosCapa1[0]-alfa*der_sesgo0(y[i],prediccion,pesosCapa2,salidascapa1,x[i]).item()
        pesosCapa1[1,0]=pesosCapa1[1,0]-alfa*der_p10(y[i],prediccion,pesosCapa2,salidascapa1,x[i]).item()
        pesosCapa1[1,1]=pesosCapa1[1,1]-alfa*der_p11(y[i],prediccion,pesosCapa2,salidascapa1,x[i]).item()
        sesgosCapa1[1]=sesgosCapa1[1]-alfa*der_sesgo1(y[i],prediccion,pesosCapa2,salidascapa1,x[i]).item()
        pesosCapa2[0]=pesosCapa2[0]-alfa*der_p0(y[i],prediccion,salidascapa1).item()
        pesosCapa2[1]=pesosCapa2[1]-alfa*der_p1(y[i],prediccion,salidascapa1).item()
        sesgosCapa2[0]=sesgosCapa2[0]-alfa*der_sesgoF(y[i],prediccion).item()

        error+=EC(y[i],prediccion)

    print(f'Epoca:{ep} Error cuadratico medio:{error/len(x)}')




print('Proceso terminado')


```
