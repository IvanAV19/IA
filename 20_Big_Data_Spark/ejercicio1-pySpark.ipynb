```python
from pyspark.sql import SparkSession

# Crear una sesión de Spark
spark = SparkSession.builder \
    .appName("EjercicioRDD") \
    .getOrCreate()

# Acceder al contexto de Spark (SparkContext)
sc = spark.sparkContext
```

```python
from google.colab import output

# Forzamos una acción para despertar la UI
spark.range(1).count()

# Generamos el enlace de acceso externo
output.serve_kernel_port_as_window(4040, path='/jobs/')
```

```python
# Crear un RDD
rdd_numeros = sc.parallelize(range(1, 21))

print(f'Elementos del RDD: {rdd_numeros.collect()}')
```

```python
# Filtrar múltiplos de 3
rdd_multiplos_3 = rdd_numeros.filter(lambda x: x % 3 == 0)

print(f'Múltiplos de 3: {rdd_multiplos_3.collect()}')
```

```python
# Crear un DataFrame sencillo
datos = [("Ana", "Madrid", 30),
("Luis", "Sevilla", 25),
("Carlos", "Valencia", 35),
("Laura", "Barcelona", 28),
("Jorge", "Madrid", 40)]

columnas = ['nombre', 'Ciudad', 'Edad']
df_personas = spark.createDataFrame(datos, schema=columnas)

df_personas.show()
```

```python
# Filtrar personas por ciudad
df_madrid = df_personas.filter(df_personas.Ciudad == 'Madrid')

df_madrid.show()
```

```python
# Agrupar personas por ciudad
df_conteo_ciudades = df_personas.groupBy('Ciudad').count()

df_conteo_ciudades.show()
```

```python
# Cerrar la sesión Spark
spark.stop()
```
