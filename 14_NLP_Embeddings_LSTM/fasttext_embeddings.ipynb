```python
import fasttext
import os
```

```python
# Preparación de datos
# Creamos un archivo de texto 'datos.txt' con oraciones de ejemplo
texto_ejemplo = """
el teclado es un periférico de entrada hardware
el ratón sirve para apuntar en la pantalla
el monitor muestra la información visual
la cpu procesa los datos del ordenador
el teclado mecánico es ruidoso pero preciso
el teclado y el ratón son dispositivos usb
la pantalla es led y tiene alta resolución
el software se ejecuta en el hardware
"""

# Guardamos este texto en un archivo físico
with open("datos_tecnologia.txt", "w", encoding="utf-8") as f:
    f.write(texto_ejemplo)
```

```python
# Entrenamiento del modelo
# model='skipgram': bueno para aprender relaciones semánticas
# minCount=1: Importante para datos pequeños (acepta palabras que aparecen solo 1 vez)
# epoch=25: Repetimos el entrenamiento varias veces porque tenemos pocos datos
model = fasttext.train_unsupervised('datos_tecnologia.txt', model='skipgram', minCount=1, epoch=25)
```

```python
# Visualizamos el vector de una palabra
vector_teclado = model.get_word_vector("teclado")

print("Dimensión del vector")
print(f"Tamaño del vector: {len(vector_teclado)}")
print(f"Primeros 10 valores del vector de 'teclado':\n{vector_teclado[:10]}")
print("\n" + "="*40 + "\n")
```

```python
# Interrogatorio interactivo
# Usamos input() para pedir la palabra por teclado
palabra_usuario = input("Introduce una palabra para buscar sus relacionadas (ej. teclado): ")

# Buscamos las 5 palabras más cercanas (k=5)
# FastText devuelve una lista de tuplas (similitud, palabra)
# Nota: get_nearest_neighbors devuelve primero la probabilidad, luego la palabra.
vecinos = model.get_nearest_neighbors(palabra_usuario, k=5)

print(f"\nPalabras más relacionadas con '{palabra_usuario}':")
for similitud, palabra in vecinos:
    print(f"- {palabra} (Similitud: {similitud:.4f})")
```
