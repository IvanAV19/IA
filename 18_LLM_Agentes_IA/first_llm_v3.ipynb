```python
from transformers import pipeline
```

```python
# Cargamos dos modelos
model_a = pipeline("text-generation", model="openai-community/gpt2")
model_b = pipeline("text-generation", model="distilgpt2")
```

```python
# Parámetros de "personalidad"
config_chat = {
    "max_new_tokens": 50,
    "do_sample": True,     # ¡Importante! Permite la creatividad
    "temperature": 0.8,    # 0.1 es muy serio, 1.0 es muy caótico
    "top_k": 50,           # Solo elige entre las 50 palabras más probables
    "top_p": 0.95,         # Nucleus sampling: ignora las palabras muy raras
    "pad_token_id": 50256, # Tomado de tu notebook
    "truncation": True
}

contexto = "The secret to a long life is"
```

```python
print(f"--- Inicio del diálogo ---\n")

for i in range(4): # 4 turnos de conversación
    # Turno del Modelo A
    res_a = model_a(contexto, **config_chat)[0]["generated_text"]
    nueva_frase_a = res_a[len(contexto):].strip().split('.')[0] + "." # Tomamos solo la primera oración nueva
    print(f"Modelo A: {nueva_frase_a}")

    # El contexto para B es lo que dijo A
    contexto = nueva_frase_a

    # Turno del Modelo B
    res_b = model_b(contexto, **config_chat)[0]["generated_text"]
    nueva_frase_b = res_b[len(contexto):].strip().split('.')[0] + "."
    print(f"Modelo B: {nueva_frase_b}\n")

    contexto = nueva_frase_b
```
