```python
from transformers import pipeline
```

```python
# Cargamos dos modelos
model_a = pipeline("text-generation", model="openai-community/gpt2")
model_b = pipeline("text-generation", model="distilgpt2")
```

```python
# Configuración inicial
contexto = "Hello! I am an AI. Let's talk about the future of humanity."
turnos = 5  # Cuántas veces quieres que intercambien mensajes
```

```python
print(f"Inicio de la conversación\n")

for i in range(turnos):
    # Turno del Modelo A
    # Generamos respuesta basándonos en el contexto actual
    out_a = model_a(contexto, max_new_tokens=40, pad_token_id=50256, truncation=True)[0]["generated_text"]

    # Extraemos solo lo nuevo
    nueva_respuesta_a = out_a[len(contexto):].strip()
    print(f"Modelo A: {nueva_respuesta_a}\n")

    # El contexto para el Modelo B es la respuesta del Modelo A
    contexto = nueva_respuesta_a

    # Turno del Modelo B
    out_b = model_b(contexto, max_new_tokens=40, pad_token_id=50256, truncation=True)[0]["generated_text"]

    nueva_respuesta_b = out_b[len(contexto):].strip()
    print(f"Modelo B: {nueva_respuesta_b}\n")

    # Actualizamos el contexto para la siguiente ronda del Modelo A
    contexto = nueva_respuesta_b
```
