```python
from transformers import pipeline
```

```python
# Cargamos dos modelos
filosofo = pipeline("text-generation", model="openai-community/gpt2")
robot_sad = pipeline("text-generation", model="distilgpt2")
```

```python
# Definimos los "perfiles"
prompt_filosofo = "I am a wise philosopher who loves life and beauty. I say: "
prompt_robot = "I am a cold, cynical robot who hates existence. I reply: "
```

```python
# Configuración de locura/creatividad
config = {
    "max_new_tokens": 40,
    "do_sample": True,
    "temperature": 0.9,
    "top_p": 0.9,
    "pad_token_id": 50256
}

# El tema inicial
charla = "The sunset is beautiful today."
```

```python
print(f"Inicio del Debate\n")
print(f"Filósofo inicial: {charla}\n")

for i in range(10):
    # Turno del Robot Pesimista
    # Le pasamos su perfil + lo que dijo el filósofo
    input_robot = f"{prompt_robot} '{charla}'"
    res_robot = robot_sad(input_robot, **config)[0]["generated_text"]

    # Limpiamos para quedarnos solo con su respuesta
    respuesta_robot = res_robot[len(input_robot):].strip().split('\n')[0]
    print(f"Robot Pesimista: {respuesta_robot}")

    # Turno del Filósofo
    input_filosofo = f"{prompt_filosofo} '{respuesta_robot}'"
    res_filosofo = filosofo(input_filosofo, **config)[0]["generated_text"]

    charla = res_filosofo[len(input_filosofo):].strip().split('\n')[0]
    print(f"Filósofo: {charla}\n")
```
