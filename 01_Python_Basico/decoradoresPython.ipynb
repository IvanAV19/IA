```python
def decorador(funcion):
    def decorada(cosa):
        print(funcion.__doc__)
        funcion(cosa)
    return decorada

@decorador
def funcioncilla(cosa):
    '''Esta herramienta hace cosas alucinantes'''
    print('He hecho una cosa alucinante')
    print(cosa)

funcioncilla('cosa')
```

```python
def numerizar_salida(funcion):
    def numerizada(*args):
        retorno = funcion(*args)
        return (tuple([float(num) for num in retorno]))
    return numerizada

@numerizar_salida
def obtener_datos(a, b, c, d):
    return (a, b, c, d)
```

```python
from dataclasses import dataclass

@dataclass
class Miclase:
    pepe = 8
    nombre: str
    edad: int
    ciudad: str

    @staticmethod
    def metodo_estatico():
        print('Este método es estático')

    def saludar(self):
        print(f'Hola, mi nombre es {self.nombre}')

instancia1 = Miclase('pepe', 25, 'Madrid')
instancia2 = Miclase('juan', 30, 'Barcelona')

print(f'Atributo estático pepe {Miclase.pepe}')
print(f'Atributo estático pepe, desde instancia1 {instancia1.pepe}')

#Miclase.pepe = 100
instancia2.pepe = 200

print(f'Atributo estático pepe, desde instancia2 {instancia2.pepe}')
print(f'Atributo estático pepe, desde instancia1 {instancia1.pepe}')
print(instancia1)

instancia1.saludar()
Miclase.metodo_estatico()
```
