```python
class MiError(Exception):
    pass

try:
    # Código que puede lanzar una excepción
    #a = 5/0
    lista = [1]

    raise MiError('Mensaje de error')

    print(lista[5])

except IndexError as ie:
    print('Indice fuera de rango')
except ZeroDivisionError as zde:
    print('No se puede dividir entre cero')
except MiError as me:
    print(me)
finally:
    print('Fin del programa')
```

```python
class ExcepcionEdad(Exception):
    def __init__(self, mensaje):
        self.mensaje = mensaje

class Persona:
    def __init__(self, nombre, dni, edad): # Método Dunder o Mágico
        if(edad <= 18 > 100):
            raise ExcepcionEdad('La edad no puede ser menor de 18 años')
        self.nombre = nombre
        self.dni = dni
        self.edad = edad

    def __str__(self):
        return f'Nombre: {self.nombre}, DNI: {self.dni}, Edad: {self.edad}'

    def incrementar_edad(self, incremento):
        self.edad += incremento

class Trabajador(Persona):
    def __init__(self, nombre, dni, edad, salario):
        super().__init__(nombre, dni, edad)
        self.salario = salario

    def __str__(self):
        return f'{super().__str__()} Salario: {self.salario}'

trabajador = Trabajador('Pepe', '12345678A', 25, 1500)
print(trabajador)
persona = Persona('Paco', '25569874P', 57)
persona.incrementar_edad(10)
#print(dir(persona)) # Muestra los métodos de la clase
print(persona)
```
