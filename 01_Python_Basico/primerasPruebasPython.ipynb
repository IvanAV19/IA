```python
lista_ht = [1, 2, 3.5, 4.7, 'gato', 'perro']

#VERSIÓN SIMPLIFICADA
nova_lista1 = [x for x in lista_ht if type(x) != str]

print(nova_lista1)

#VERSIÓN NORMAL
nova_lista2 = []

for x in lista_ht:
    if type(x) != str:
        nova_lista2.append(x)

print(nova_lista2)
```

```python
def factorial(n):
    if n == 1:
        return n
    else:
        return n*factorial(n - 1)

print(factorial(3))
```

```python
def me_vale_todo(*args, **kwargs):
    print(args)
    print(kwargs)
    print(f'El primer argumento posicional es: {args[0]}')
    print(f'El argumento a es: {kwargs["a"]}')

me_vale_todo(1, 2, 3, a = 4.0, b = 45.9, d = 'pepe')
```
