Definir una función que usando map, reciba 4 listas y obtenga como salida una lista cuyos valores sean tuplas formadas por los elementos provenientes de dichas listas.

```python
lista1 = [1, 2, 3, 4]
lista2 = ['a', 'b', 'c', 'd']
lista3 = [True, False, True, False]
lista4 = [1.2, 3.4, 5.6, 7.8]

def combinar_listas_con_map(list1, list2, list3, list4):
  """
  Combina cuatro listas en una lista de tuplas usando map()

  Argumentos:
    list1, list2, list3, list4: Las listas a combinar

  Salida:
    Una lista de tuplas, donde cada tupla contiene elementos de las listas de entrada
  """
  return list(map(lambda a, b, c, d: (a, b, c, d), list1, list2, list3, list4))

listaFinal = combinar_listas_con_map(lista1, lista2, lista3, lista4)
print(listaFinal)
```

Crear una clase Tabla con un diccionario como almacen de columnas y filas. Las claves del diccionario serán los nombres de las columnas y para cada columna tendremos una lista con todos los valores de los datos. El control del número de filas lo haremos mediante un atributo entero que registra el número de filas actual. Deberá modificarse con cualquier operación que altere la tabla. Los métodos a desarrollar serán:

El constructor
Añadir una nueva columna. Si ya hay datos la lista que se corresponde con esa columna tendrá tantos nulos como filas haya.
Añadir una nueva fila a la tabla.
Suprimir una fila.
Encontrar una fila de datos en función de un valor de una determinada columna por ejemplo encontrar_fila('nif', '45896574J')

```python
class Tabla:
    def __init__(self):
        # Diccionario para guardar las columnas y sus listas de datos
        self.columnas = {}
        # Contador de filas actuales
        self.num_filas = 0

    def anhadir_columna(self, nombre_columna):
        # Si la columna ya existe devolvemos un aviso
        if nombre_columna in self.columnas:
            print(f"Error: La columna '{nombre_columna}' ya existe.")
            return # El return vacío hace que la función termine aquí

        # Si ya hay filas, rellenamos la nueva columna con valores nulos (None)
        if self.num_filas > 0:
            self.columnas[nombre_columna] = [None] * self.num_filas
        else:
            self.columnas[nombre_columna] = []

    def anhadir_fila(self, datos):
        """
        datos: diccionario con los valores de la nueva fila
        ejemplo: {'nombre': 'Ana', 'edad': 25}
        """
        # Añadimos valores a cada columna
        for columna in self.columnas:
            if columna in datos:
                self.columnas[columna].append(datos[columna])
            else:
                self.columnas[columna].append(None)
        self.num_filas += 1

    def suprimir_fila(self, indice):
        # Elimina la fila con el número indicado (empezando en 0)
        if 0 <= indice < self.num_filas:
            for columna in self.columnas:
                del self.columnas[columna][indice]
            self.num_filas -= 1
        else:
            print("Índice de fila fuera de rango")

    def encontrar_fila(self, nombre_columna, valor):
        # Busca una fila donde la columna tenga el valor indicado
        if nombre_columna not in self.columnas:
            print("La columna no existe")
            return None

        for i, v in enumerate(self.columnas[nombre_columna]):
            if v == valor:
                # Construimos un diccionario con los valores de esa fila
                fila = {}
                for col in self.columnas:
                    fila[col] = self.columnas[col][i]
                return fila
        print("No se encontró la fila")
        return None


# ==== Pruebas ====
tabla = Tabla()
tabla.anhadir_columna("nombre")
tabla.anhadir_columna("edad")
tabla.anhadir_columna("nif")

tabla.anhadir_fila({"nombre": "Ana", "edad": 25, "nif": "45896574J"})
tabla.anhadir_fila({"nombre": "Luis", "edad": 30, "nif": "12345678A"})

print('Buscamos la fila con nif 45896574J')
print(tabla.encontrar_fila("nif", "45896574J"))
print()

print('Columnas antes de eliminar una:')
print(tabla.columnas)
print()

tabla.suprimir_fila(0)

print('Columnas después de eliminar una:')
print(tabla.columnas)
```
