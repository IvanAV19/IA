Vamos a desarrollar un decorador que permita la depuración de funciones.
Cuando la función a decorar tenga parámetros de entrada los guardaremos en un archivo, haremos lo mismo con las salidas de la función a decorar.

```python
def guardar_entradas(*args, **kwargs):
    with open('entradas.txt', 'a') as f:
        for arg in args:
            f.write(f'{arg}\n')

        for k, v in kwargs.items():
            f.write(f'{k}={v}\n')

def depurador(funcion):
    def decorador(*args, **kwargs):
        guardar_entradas(f'Nombre de la función: {funcion.__name__} - Argumentos de entrada: ', *args, **kwargs)
        resultado = funcion(*args, **kwargs)
        guardar_entradas('Resultado: ', resultado)

    return decorador

@depurador
def funcion(a, b, mensaje = 'No hay mensaje'):
    return a + b

for i in range(10):
    funcion(100, 200, mensaje = 'Hola chaval')
```
