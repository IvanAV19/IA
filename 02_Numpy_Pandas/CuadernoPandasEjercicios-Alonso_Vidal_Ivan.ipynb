##5.- Obten el dataset

##Efectúa un join entre este dataframe y el de drivers por el atributo driverId

```python
df_standings = pd.read_csv('/content/drive/MyDrive/Archivos .csv/driverStandings.csv')
dfidx1 = df.set_index('driverId', inplace = False)
dfidx2 = df_standings.set_index('driverId', inplace = False)
df_join = dfidx1.join(dfidx2, how = 'inner')
df_join
```

##6.- Mediante groupby suma los puntos de cada corredor, de forma que aparezca el apellido del corredor

```python
df_puntos = df_join.groupby(['driverId', 'surname'])['points'].sum()
df_puntos
```

##7.- Del dataframe del ejercicio 6 suprime la columna url, para obtener un nuevo dataframe

```python
df_sin_url = df_join.drop(columns = 'url')
df_sin_url.head()
```

##8.- Con el dataframe anterior genera uno nuevo en el que el apellido aparezca como cabecera de columna y aparezca la posición en las carreras en las que ha participado (puedes usar raceId y driverStandingsId como indices). Luego extrae las columnas correspondientes a Alonso , Hamilton y Rosberg

```python
df_surname = df_sin_url.pivot(index = ['raceId', 'driverStandingsId'], columns='surname', values = 'position')
df_surname.head()
df_surname[['Alonso', 'Hamilton', 'Rosberg']]
```

##9.-Con el dataframe del ejercicio 7, añadirle tres columnas con valores enteros aleatorios con nombres col1, col2 y col3

```python
df_sin_idx = df_sin_url.reset_index()
nuevodf = pd.DataFrame(np.random.rand(df_sin_idx.shape[0], 3), columns = ['col1', 'col2', 'col3'])

df_concat = pd.concat([df_sin_idx, nuevodf], axis = 1, )
df_concat.head()
```
